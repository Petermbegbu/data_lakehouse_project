#### Read files from the volume source and 
#### write to a delta table in the bronze schema

```python
from modules.path_list import ERP_PATH, CRM_PATH

# Write a function that reads and writes all files from volume to bronze catalog

def read_and_write_crm():
  for file in dbutils.fs.ls(CRM_PATH):
    #read file
    df = spark.read.format("csv").options(header="true", inferSchema="true").load(file.path)
    
    #write file
    df.write.mode("overwrite").saveAsTable("data_lakehouse.bronze.crm_"+file.name.replace(".csv",""))

def read_and_write_erp():
  for file in dbutils.fs.ls(ERP_PATH):
    #read file
    df = spark.read.format("csv").options(header="true", inferSchema="true").load(file.path)
  
    #write file
    df.write.mode("overwrite").saveAsTable("data_lakehouse.bronze.erp_"+file.name.replace(".csv",""))  

# Read and write all files from volume to bronze catalog
read_and_write_crm()
read_and_write_erp()
    
print("All files have been read and written to the bronze catalog.")
```
